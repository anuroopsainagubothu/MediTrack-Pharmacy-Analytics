```python
sales_drugs = set(sales_clean['DrugCode'].dropna())

stock_drugs = set(stock_clean['DrugCode'].dropna())

common_drugs = sales_drugs.intersection(stock_drugs)

print("Unique Sales DrugCodes:", len(sales_drugs))

print("Unique Stock DrugCodes:", len(stock_drugs))

print("Common DrugCodes:", len(common_drugs))

print("Sales-only DrugCodes:", len(sales_drugs - stock_drugs))

print("Stock-only DrugCodes:", len(stock_drugs - sales_drugs))
```

```python
def get_unique_value(series):
    values = series.dropna().unique()

    if len(values) == 1:
        return values[0]

    return pd.NA


stock_lookup = stock_clean.groupby('DrugCode').agg(

    DrugClass_Count=('DrugClass',
                     lambda x: x.dropna().nunique()),

    DrugClass_Value=('DrugClass',
                     get_unique_value),

    Category_Count=('Category',
                    lambda x: x.dropna().nunique()),

    Category_Value=('Category',
                    get_unique_value)

).reset_index()

display(stock_lookup.head(10))
```

```python
print("DrugClass conflicts:",
      (stock_lookup['DrugClass_Count'] > 1).sum())

print("Category conflicts:",
      (stock_lookup['Category_Count'] > 1).sum())
```

```python
drugclass_map = stock_lookup.set_index(
    'DrugCode'
)['DrugClass_Value']

category_map = stock_lookup.set_index(
    'DrugCode'
)['Category_Value']

sales_clean['DrugClass_from_stock'] = (
    sales_clean['DrugCode'].map(drugclass_map)
)

sales_clean['Category_from_stock'] = (
    sales_clean['DrugCode'].map(category_map)
)

print("Recoverable DrugClass records:",
      (sales_clean['DrugClass'].isna() &
       sales_clean['DrugClass_from_stock'].notna()).sum())

print("Recoverable Category records:",
      (sales_clean['Category'].isna() &
       sales_clean['Category_from_stock'].notna()).sum())
```
