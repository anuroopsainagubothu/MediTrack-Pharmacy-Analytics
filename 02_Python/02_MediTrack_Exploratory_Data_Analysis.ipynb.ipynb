### 6.1 Top 10 Medicines by Revenue

This analysis identifies the ten medicines generating the highest total revenue.

```python
display(
    sales.nlargest(10, 'DiscountAmount')[[
        'InvoiceNumber',
        'DrugName',
        'Quantity',
        'UnitPrice',
        'DiscountPct',
        'DiscountAmount',
        'Subtotal',
        'TotalPrice'
    ]]
)
```

```python
category_sales = (
    sales.groupby('Category')
    .agg(
        Total_Revenue=('TotalPrice', 'sum'),
        Total_Quantity=('Quantity', 'sum'),
        Total_Transactions=('InvoiceNumber', 'nunique')
    )
    .sort_values('Total_Revenue', ascending=False)
)

display(category_sales)
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=category_sales.reset_index(),
    x='Total_Revenue',
    y='Category'
)

plt.title('Total Revenue by Pharmacy Category')
plt.xlabel('Total Revenue')
plt.ylabel('Product Category')

plt.tight_layout()
plt.show()
```

```python
category_sales['Revenue_Percentage'] = (
    category_sales['Total_Revenue']
    / category_sales['Total_Revenue'].sum()
) * 100

display(
    category_sales[
        ['Total_Revenue', 'Revenue_Percentage']
    ].round(2)
)
```

```python
top_10_medicines = (
    sales.groupby('DrugName')
    .agg(
        Total_Revenue=('TotalPrice', 'sum'),
        Total_Quantity=('Quantity', 'sum')
    )
    .sort_values('Total_Revenue', ascending=False)
    .head(10)
)

display(top_10_medicines)
```

```python
top_10_chart = top_10_medicines.reset_index()

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_chart,
    x='Total_Revenue',
    y='DrugName'
)

plt.title('Top 10 Medicines by Revenue')
plt.xlabel('Total Revenue')
plt.ylabel('Medicine Name')

plt.tight_layout()
plt.show()
```

```python
top_10_quantity = (
    sales.groupby('DrugName')
    .agg(
        Total_Quantity=('Quantity', 'sum'),
        Total_Revenue=('TotalPrice', 'sum')
    )
    .sort_values('Total_Quantity', ascending=False)
    .head(10)
)

display(top_10_quantity)
```

## This analysis identifies the Expired and NON-Expired Sales 

It helps understand product movement and identify medicines with relatively Expired and NON-Expired

The analysis complements revenue-based product rankings by focusing on quantity rather than monetary value.

```python
# Identify expired and non-expired medicines
reference_date = pd.Timestamp('2025-12-31')

sales['Expiry_Status'] = np.where(
    sales['ExpiryDate'].isna(),
    'Missing Expiry Date',
    np.where(
        sales['ExpiryDate'] < reference_date,
        'Expired',
        'Not Expired'
    )
)

display(sales['Expiry_Status'].value_counts())
```

```python
sales['Sold_After_Expiry'] = (
    sales['DateTime'].dt.date >
    sales['ExpiryDate'].dt.date
)

display(sales['Sold_After_Expiry'].value_counts(dropna=False))
```

```python
expired_sales = sales[
    sales['Sold_After_Expiry'] == True
]

display(
    expired_sales[
        ['DateTime', 'DrugCode', 'DrugName',
         'BatchNumber', 'ExpiryDate', 'Quantity', 'TotalPrice']
    ].head(20)
)
```

```python
expired_product_analysis = (
    expired_sales.groupby(['DrugCode', 'DrugName'])
    .agg(
        Expired_Sale_Records=('DrugCode', 'count'),
        Total_Quantity=('Quantity', 'sum'),
        Total_Revenue=('TotalPrice', 'sum')
    )
    .sort_values(
        'Expired_Sale_Records',
        ascending=False
    )
)

display(expired_product_analysis.head(15))
```

```python
top_10_expired = (
    expired_product_analysis
    .head(10)
    .reset_index()
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_10_expired,
    x='Expired_Sale_Records',
    y='DrugName'
)

plt.title('Top 10 Medicines in Sales Records After Recorded Expiry')
plt.xlabel('Number of Flagged Sales Records')
plt.ylabel('Medicine Name')

plt.tight_layout()
plt.show()
```

### Expiry Analysis – Key Findings

- L Salep A had the highest number of flagged records (733).
- Bufamoxy 500 ranked second (306 records).
- These records require verification against original batch and expiry information.

## 4. Importing Data into MySQL Using Python

The cleaned datasets were inserted into MySQL using `executemany()` from MySQL Connector/Python.

To handle large datasets efficiently, records were imported in batches of 5,000 rows.

The following steps were performed:

1. Prepared SQL INSERT statements.
2. Converted Pandas and NumPy values into MySQL-compatible values.
3. Handled missing values using NULL.
4. Inserted records in batches of 5,000.
5. Committed each batch using `commit()`.

The Sales dataset was imported successfully with 649,656 records, followed by the Stock dataset with 804,027 records.

```python
import numpy as np
import pandas as pd

# Prepare SQL query
columns = list(sales_df.columns)

placeholders = ", ".join(["%s"] * len(columns))
column_names = ", ".join(f"`{col}`" for col in columns)

insert_query = f"""
INSERT INTO sales ({column_names})
VALUES ({placeholders})
"""

# Convert pandas/numpy values into MySQL-compatible values
def convert_value(value):
    if pd.isna(value):
        return None
    if isinstance(value, pd.Timestamp):
        return value.to_pydatetime()
    if isinstance(value, np.generic):
        return value.item()
    return value

# Import in batches
batch_size = 5000
total_rows = len(sales_df)

for start in range(0, total_rows, batch_size):
    batch = sales_df.iloc[start:start + batch_size]

    records = [
        tuple(convert_value(value) for value in row)
        for row in batch.itertuples(index=False, name=None)
    ]

    cursor.executemany(insert_query, records)
    conn.commit()

    print(f"Imported {min(start + batch_size, total_rows):,} / {total_rows:,}")

print("Sales data imported successfully!")
```
